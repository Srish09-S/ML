```python
import seaborn as sns
tips=sns.load_dataset('tips')
tips
```

```python
sns.regplot(data=tips , x='tip' , y='total_bill')
```

```python
sns.lmplot(data=tips , x='tip' , y='total_bill' , hue='sex')
```

```python
sns.residplot(data=tips , x='tip' , y='total_bill')
```
