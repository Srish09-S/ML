```python
import seaborn as sns
```

```python
tips=sns.load_dataset("tips")
```

```python
tips.head()
```

```python
sns.stripplot(data=tips, x='sex' , y='total_bill')
```

```python
sns.scatterplot(data=tips , x='day' , y='total_bill')
```

```python
sns.stripplot(data=tips, x='sex' , y='total_bill' , jitter=False)
```

```python
sns.stripplot(data=tips, x='sex' , y='total_bill' , hue='sex')
```

```python
sns.catplot(data=tips, x='sex' , y='total_bill' , kind='strip')
```

```python
sns.catplot(data=tips, x='sex' , y='total_bill' , kind='strip' , jitter=0.3)
```

```python
sns.catplot(data=tips, x='sex' , y='total_bill' , kind='strip' , hue='sex')
```

```python
sns.swarmplot(data=tips, x='sex' , y='total_bill')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='swarm')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='swarm',hue='sex')
```

```python
sns.boxplot(data=tips,x='sex',y='total_bill')
```

```python
sns.boxplot(data=tips,x='day',y='total_bill')
```

```python
sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
sns.boxplot(data=tips,y='total_bill')
```

```python
sns.catplot(data=tips, x='day' , y='total_bill' , hue='sex' , kind='box')
```

```python
sns.violinplot(data=tips , x='day' , y='total_bill' )
```

```python
sns.violinplot(data=tips , x='day' , y='total_bill' , hue='sex')
```
